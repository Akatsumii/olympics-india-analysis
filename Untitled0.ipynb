# Why Does India Win Fewer Olympic Medals Than Its Population Suggests?
### A Data Analysis of Olympic Performance (1896-2024)

## Part 1: Data collection
All data is loaded by code, with no manual files.
- Olympic history 1896-2016 (Kaggle)
- Paris 2024 medal tables (Kaggle)
- Tokyo 2020 medal tables (Kaggle)
- Population and GDP, 1960-2024 (World Bank API)

```python
!pip install kagglehub openpyxl -q

import os, glob, requests, itertools
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import kagglehub
import warnings
warnings.filterwarnings('ignore')

def load_kaggle(slug):
    folder = kagglehub.dataset_download(slug)
    data = {}
    for f in glob.glob(f"{folder}/**/*", recursive=True):
        name = os.path.basename(f)
        if f.lower().endswith(".csv"):
            try:
                data[name] = pd.read_csv(f)
            except UnicodeDecodeError:
                data[name] = pd.read_csv(f, encoding="latin-1")
        elif f.lower().endswith((".xlsx", ".xls")):
            data[name] = pd.read_excel(f)
    return data

def load_first(slugs):
    for s in slugs:
        try:
            d = load_kaggle(s)
            if d:
                print("Loaded:", s)
                return d
        except Exception as e:
            print("Could not load", s, "->", type(e).__name__)
    raise RuntimeError("None of these datasets loaded: " + ", ".join(slugs))

def world_bank(indicator, name):
    url = f"https://api.worldbank.org/v2/country/all/indicator/{indicator}"
    all_data = []
    page = 1
    while True:
        r = requests.get(url, params={"format": "json", "per_page": 1000, "page": page, "date": "1960:2024"}, timeout=60)
        response_json = r.json()
        if len(response_json) < 2:
            break
        rows = response_json[1]
        if not rows:
            break
        all_data.extend(rows)
        # Read total pages from metadata
        total_pages = response_json[0].get("pages", 1)
        if page >= total_pages:
            break
        page += 1
    return pd.DataFrame([{"country": d["country"]["value"],
                          "iso3": d["countryiso3code"],
                          "year": int(d["date"]),
                          name: d["value"]} for d in all_data])

olympics = load_first(["mysarahmadbhat/120-years-of-olympic-history",
                       "heesoo37/120-years-of-olympic-history-athletes-and-results"])
paris = load_first(["piterfm/paris-2024-olympic-summer-games"])
tokyo = load_first(["arjunprasadsarkhel/2021-olympics-in-tokyo",
                    "piterfm/tokyo-2020-olympics"])
pop = world_bank("SP.POP.TOTL", "population")
gdp = world_bank("NY.GDP.MKTP.CD", "gdp_usd")

print("\nData loaded.")
```

```python
# ---------- Athlete data: Summer Games, one medal per team per event ----------
ath = olympics['athlete_events.csv']
summer = ath[(ath['Season'] == 'Summer') & (ath['Sport'] != 'Art Competitions')].copy()
medal_rows = summer[summer['Medal'].notna()]
medals_clean = medal_rows.drop_duplicates(subset=['Year', 'NOC', 'Event', 'Medal']).copy()

medal_counts = (medals_clean.groupby(['Year', 'NOC', 'Medal']).size()
                .unstack(fill_value=0).reset_index())
medal_counts.columns.name = None
medal_counts['Total'] = medal_counts[['Gold', 'Silver', 'Bronze']].sum(axis=1)
medal_counts = medal_counts[['Year', 'NOC', 'Gold', 'Silver', 'Bronze', 'Total']]

# ---------- World Bank: keep real countries only ----------
c = requests.get("https://api.worldbank.org/v2/country",
                 params={"format": "json", "per_page": 500}, timeout=60).json()[1]
countries = pd.DataFrame([{"iso3": x["id"], "region": x["region"]["value"]} for x in c])
real_iso3 = set(countries.loc[countries['region'] != 'Aggregates', 'iso3'])
wb = pop.merge(gdp[['iso3', 'year', 'gdp_usd']], on=['iso3', 'year'], how='inner')
wb = wb[wb['iso3'].isin(real_iso3)].copy()

# ---------- Paris 2024 ----------
paris_raw = paris['medals_total.csv']
paris_std = pd.DataFrame({
    'Year': 2024, 'NOC': paris_raw['country_code'],
    'Country': paris_raw['country_long'],
    'Gold': paris_raw['Gold Medal'], 'Silver': paris_raw['Silver Medal'],
    'Bronze': paris_raw['Bronze Medal'], 'Total': paris_raw['Total'],
}).dropna(subset=['NOC'])

# ---------- Tokyo 2020 ----------
tokyo_raw = tokyo['Medals.xlsx']
name_to_noc = dict(zip(paris_raw['country_long'], paris_raw['country_code']))
name_to_noc.update(dict(zip(paris_raw['country'], paris_raw['country_code'])))
manual = {
    "ROC": "RUS", "Republic of Korea": "KOR", "Islamic Republic of Iran": "IRI",
    "Chinese Taipei": "TPE", "Hong Kong, China": "HKG", "Türkiye": "TUR",
    "Turkey": "TUR", "Côte d'Ivoire": "CIV", "Czech Republic": "CZE",
    "Republic of Moldova": "MDA", "Syrian Arab Republic": "SYR",
    "Kosovo": "KOS", "Great Britain": "GBR",
}
tokyo_std = pd.DataFrame({
    'Year': 2020, 'Country': tokyo_raw['Team/NOC'],
    'Gold': tokyo_raw['Gold'], 'Silver': tokyo_raw['Silver'],
    'Bronze': tokyo_raw['Bronze'], 'Total': tokyo_raw['Total'],
})
tokyo_std['NOC'] = tokyo_std['Country'].map(lambda n: name_to_noc.get(n, manual.get(n)))
missing = tokyo_std[tokyo_std['NOC'].isna()]
print("Tokyo countries without a code:", len(missing), missing['Country'].tolist())

# ---------- Combine all Games ----------
cols = ['Year', 'NOC', 'Gold', 'Silver', 'Bronze', 'Total']
recent = pd.concat([tokyo_std[cols], paris_std[cols]], ignore_index=True).dropna(subset=['NOC'])
all_medals = pd.concat([medal_counts, recent], ignore_index=True)

# ---------- Olympic code -> World Bank code ----------
noc_to_iso3 = {
    "GER": "DEU", "NED": "NLD", "SUI": "CHE", "DEN": "DNK", "GRE": "GRC",
    "POR": "PRT", "SLO": "SVN", "CRO": "HRV", "BUL": "BGR", "RSA": "ZAF",
    "UAE": "ARE", "KSA": "SAU", "IRI": "IRN", "CHI": "CHL", "URU": "URY",
    "INA": "IDN", "MAS": "MYS", "PHI": "PHL", "SIN": "SGP", "VIE": "VNM",
    "BAN": "BGD", "NEP": "NPL", "SRI": "LKA", "MGL": "MNG", "LAT": "LVA",
    "ALG": "DZA", "NGR": "NGA", "ZAM": "ZMB", "ZIM": "ZWE", "TAN": "TZA",
    "LBA": "LBY", "LIB": "LBN", "KUW": "KWT", "OMA": "OMN", "BRN": "BHR",
    "BAH": "BHS", "BAR": "BRB", "HAI": "HTI", "GUA": "GTM", "HON": "HND",
    "CRC": "CRI", "ESA": "SLV", "NCA": "NIC", "PAR": "PRY", "ANG": "AGO",
    "BOT": "BWA", "MRI": "MUS", "MAD": "MDG", "SUD": "SDN", "CGO": "COG",
    "GAM": "GMB", "GUI": "GIN", "TOG": "TGO", "BUR": "BFA", "NIG": "NER",
    "MTN": "MRT", "CHA": "TCD", "GEQ": "GNQ", "SEY": "SYC", "FIJ": "FJI",
    "SAM": "WSM", "TGA": "TON", "VAN": "VUT", "GRN": "GRD", "MYA": "MMR",
    "CAM": "KHM", "BHU": "BTN", "MON": "MCO", "KOS": "XKX", "ISV": "VIR",
    "IVB": "VGB", "BER": "BMU", "CAY": "CYM", "ARU": "ABW", "ANT": "ATG",
    "SKN": "KNA", "ASA": "ASM",
    "PUR": "PRI", "VIN": "VCT", "GBS": "GNB", "MAW": "MWI",
    "SOL": "SLB", "BIZ": "BLZ", "PLE": "PSE", "BRU": "BRN",
}
all_medals['iso3'] = all_medals['NOC'].map(lambda x: noc_to_iso3.get(x, x))

# ---------- Checks ----------
print("Duplicate Year+NOC rows:", all_medals.duplicated(['Year', 'NOC']).sum())
print("\nIndia:")
print(all_medals[(all_medals['NOC'] == 'IND') & (all_medals['Year'] >= 2008)]
      [['Year', 'NOC', 'Gold', 'Silver', 'Bronze', 'Total']])
```

## Part 2: Cleaning and integration
1. Kept Summer Games only and removed Art Competitions
2. Counted one medal per team per event (team members repeat in the raw data)
3. Kept real countries only in the World Bank data (removed regional groups)
4. Added country codes to the Tokyo 2020 table and standardized Paris 2024
5. Combined all Games into one table, `all_medals`
6. Mapped Olympic codes to World Bank codes

**Validation:** India's totals match the official results: 2008 = 3, 2012 = 6, 2016 = 2, 2020 = 7, 2024 = 6.

```python
peers = {'IND': 'India', 'CHN': 'China', 'BRA': 'Brazil',
         'INA': 'Indonesia', 'KEN': 'Kenya', 'JPN': 'Japan'}

games_years = sorted(y for y in all_medals['Year'].unique() if y >= 1960)
grid = pd.DataFrame(list(itertools.product(games_years, peers.keys())),
                    columns=['Year', 'NOC'])

mg = grid.merge(all_medals[['Year', 'NOC', 'Gold', 'Silver', 'Bronze', 'Total']],
                on=['Year', 'NOC'], how='left')
medal_cols = ['Gold', 'Silver', 'Bronze', 'Total']
mg[medal_cols] = mg[medal_cols].fillna(0).astype(int)
mg['Country'] = mg['NOC'].map(peers)
mg['iso3'] = mg['NOC'].map(lambda x: noc_to_iso3.get(x, x))

analysis = mg.merge(wb[['iso3', 'year', 'population', 'gdp_usd']],
                    left_on=['iso3', 'Year'], right_on=['iso3', 'year'],
                    how='left').drop(columns='year')
analysis = analysis.sort_values(['NOC', 'Year']).reset_index(drop=True)

analysis['pop_filled'] = analysis['population'].isna()
analysis['gdp_filled'] = analysis['gdp_usd'].isna()
analysis[['population', 'gdp_usd']] = (analysis.groupby('NOC')[['population', 'gdp_usd']].ffill())

analysis['medals_per_million'] = analysis['Total'] / (analysis['population'] / 1e6)
analysis['medals_per_billion_gdp'] = analysis['Total'] / (analysis['gdp_usd'] / 1e9)
analysis['million_people_per_medal'] = (analysis['population'] / 1e6) / analysis['Total']
analysis.loc[analysis['Total'] == 0, 'million_people_per_medal'] = np.nan

print("Games:", games_years)
print("Still missing:", analysis[['population', 'gdp_usd']].isnull().sum().to_dict())
print("\nTotal medals since 1960 per country (none should be 0):")
print(analysis.groupby('Country')['Total'].sum())
print("\n2020 and 2024 check:")
print(analysis[(analysis['Year'].isin([2020, 2024])) & (analysis['NOC'].isin(['IND', 'CHN', 'JPN']))]
      [['Year', 'NOC', 'Total']])
```

## Part 3: Analysis columns
- Peer countries: India, China, Brazil, Indonesia, Kenya, Japan
- Built a complete Games x country table (1960-2024), with 0 medals where a country won none
- Missing population or GDP for the latest year was filled with the most recent earlier value
- Created three measures:
  - Medals per million people
  - Medals per billion US$ of GDP
  - Million people per medal

```python
latest_year = analysis['Year'].max()
latest = (analysis[analysis['Year'] == latest_year]
          .sort_values('medals_per_million', ascending=False))
print("Latest Games:", latest_year)
display(latest[['Country', 'Total', 'population', 'medals_per_million',
                'medals_per_billion_gdp', 'million_people_per_medal']].round(3))

recent_games = analysis[analysis['Year'] >= 2004]
summary = (recent_games.groupby('Country')
           .agg(total_medals=('Total', 'sum'),
                avg_medals_per_million=('medals_per_million', 'mean'),
                avg_medals_per_billion_gdp=('medals_per_billion_gdp', 'mean'))
           .sort_values('avg_medals_per_million', ascending=False).round(4))
print("\nAverage over the last six Games (2004-2024):")
display(summary)
```

## Step 2: India's rank at each Games
Ranked all medal-winning countries by total medals at every Games from 1960, and extracted India's position.

**Note:** the rank is among countries that won at least one medal. Games where India won nothing show no rank.

```python
r = all_medals[all_medals['Year'] >= 1960].copy()
r['rank'] = r.groupby('Year')['Total'].rank(method='min', ascending=False)

n_countries = r.groupby('Year')['NOC'].count().rename('countries_with_medals')
india_rank = (r[r['NOC'] == 'IND'][['Year', 'Total', 'rank']]
              .set_index('Year').join(n_countries, how='right')
              .rename(columns={'Total': 'India_medals', 'rank': 'India_rank'}))
india_rank['India_medals'] = india_rank['India_medals'].fillna(0).astype(int)
india_rank
```

## Step 3: India vs. the 10 most populous countries
Total medals and medals per million people for 2004-2024, for the ten countries with the largest populations. This compares India with countries of a similar size, which is a fairer test than comparing it with small countries.

**Note:** medals per million uses the 2022 population, so it is an approximation for the whole period.

```python
POP_YEAR = 2022

top_pop = (wb[wb['year'] == POP_YEAR].dropna(subset=['population'])
           .sort_values('population', ascending=False).head(10)
           [['iso3', 'country', 'population']])

m = (all_medals[all_medals['Year'] >= 2004].groupby('iso3')['Total'].sum()
     .rename('medals_2004_2024'))

big = top_pop.merge(m, on='iso3', how='left').fillna({'medals_2004_2024': 0})
big['population_millions'] = (big['population'] / 1e6).round(0).astype(int)
big['medals_2004_2024'] = big['medals_2004_2024'].astype(int)
big['medals_per_million'] = (big['medals_2004_2024'] / (big['population'] / 1e6)).round(3)

big = big.sort_values('medals_per_million', ascending=False)
big[['country', 'population_millions', 'medals_2004_2024', 'medals_per_million']]
```

```python
d = big.sort_values('medals_per_million')
plt.figure(figsize=(8, 5))
plt.barh(d['country'], d['medals_per_million'],
         color=['#C0504D' if c == 'India' else '#1F3A5F' for c in d['country']])
plt.title("Medals per million people, 2004-2024\n(10 most populous countries)")
plt.xlabel("Medals per million people")
plt.tight_layout()
plt.show()
```

# Day 8: Statistics Dataset and Efficiency Comparison

## Questions
1. Can we build one clean table of all countries for statistical analysis?
2. How many medals does India win per athlete sent, compared with its peers?
3. How does India compare with countries of similar income per person?

**Data note:** this analysis covers Summer Games from 1960 to 2016, because athlete counts only exist in the athlete file.

## Step 1: Build the all-country panel
Created one row per country per Games (1960-2016) with medals, athletes sent (team size), population and GDP. Rows with missing population or GDP are set aside, and the number lost is reported. This table is the input for the correlation and regression on Days 9-10.

```python
team = (summer[summer['Year'] >= 1960]
        .groupby(['Year', 'NOC'])['ID'].nunique()
        .rename('athletes_sent').reset_index())

base = medal_counts[medal_counts['Year'] >= 1960][['Year', 'NOC', 'Total']]
panel = team.merge(base, on=['Year', 'NOC'], how='left')
panel['Total'] = panel['Total'].fillna(0).astype(int)

panel['iso3'] = panel['NOC'].map(lambda x: noc_to_iso3.get(x, x))
panel = panel.merge(wb[['iso3', 'year', 'population', 'gdp_usd']],
                    left_on=['iso3', 'Year'], right_on=['iso3', 'year'],
                    how='left').drop(columns='year')

panel_clean = panel.dropna(subset=['population', 'gdp_usd']).copy()
panel_clean['gdp_pc'] = panel_clean['gdp_usd'] / panel_clean['population']
panel_clean['log_pop'] = np.log(panel_clean['population'])
panel_clean['log_gdp'] = np.log(panel_clean['gdp_usd'])

print("Rows in panel         :", len(panel))
print("Rows with pop and GDP :", len(panel_clean))
print("Rows set aside        :", len(panel) - len(panel_clean))
print("Games covered         :", sorted(panel_clean['Year'].unique()))
print("\nChecks for 2016:")
panel[(panel['Year'] == 2016) & (panel['NOC'].isin(['IND', 'USA', 'CHN']))] \
    [['NOC', 'athletes_sent', 'Total', 'population', 'gdp_usd']]
```

## Step 2: Medals per 100 athletes
Compared how many medals each peer country wins for every 100 athletes it sends, from 1996 to 2016. A low value means a large team with few medals, while a high value means a small, well-targeted team.

```python
pp = panel[(panel['NOC'].isin(peers.keys())) & (panel['Year'] >= 1996)].copy()
pp['Country'] = pp['NOC'].map(peers)
pp['medals_per_100_athletes'] = (pp['Total'] / pp['athletes_sent'] * 100).round(1)

print("Athletes sent:")
display(pp.pivot(index='Year', columns='Country', values='athletes_sent'))
print("\nMedals per 100 athletes:")
display(pp.pivot(index='Year', columns='Country', values='medals_per_100_athletes'))
```

```python
piv = pp.pivot(index='Year', columns='Country', values='medals_per_100_athletes')
plt.figure(figsize=(10, 5))
for c in piv.columns:
    plt.plot(piv.index, piv[c], marker='o', label=c,
             linewidth=3 if c == 'India' else 1.5)
plt.title("Medals per 100 athletes sent, 1996-2016")
plt.xlabel("Year")
plt.ylabel("Medals per 100 athletes")
plt.legend()
plt.tight_layout()
plt.show()
```

## Step 3: India vs. countries of similar income
Selected countries whose GDP per person in 2016 was between half and double India's, then compared their medals per million people. This asks whether India underperforms even among countries with similar income.

```python
p16 = panel_clean[panel_clean['Year'] == 2016].copy()
p16['medals_per_million'] = p16['Total'] / (p16['population'] / 1e6)

india_pc = p16.loc[p16['NOC'] == 'IND', 'gdp_pc'].iloc[0]
similar = p16[(p16['gdp_pc'] >= india_pc * 0.5) & (p16['gdp_pc'] <= india_pc * 2)].copy()
similar['rank_medals_per_million'] = similar['medals_per_million'].rank(ascending=False, method='min')
similar['rank_total_medals'] = similar['Total'].rank(ascending=False, method='min')

print("India GDP per person 2016: US$", round(india_pc))
print("Countries in the similar-income group:", len(similar))
print("India's rank in the group, total medals:",
      int(similar.loc[similar['NOC'] == 'IND', 'rank_total_medals'].iloc[0]))
print("India's rank in the group, medals per million:",
      int(similar.loc[similar['NOC'] == 'IND', 'rank_medals_per_million'].iloc[0]))

display(similar.sort_values('population', ascending=False).head(10)
        [['NOC', 'population', 'gdp_pc', 'athletes_sent', 'Total', 'medals_per_million']].round(2))
```

# Day 9: Statistical Analysis

## Questions
1. Which factors are most strongly linked to medal counts?
2. How much of the variation in medals do population and GDP explain?
3. Does team size add explanatory power?
4. How many medals would we expect India to win for its population and GDP, and how far is the actual figure from that?

**Data note:** the analysis uses the all-country panel from Day 8 (1960-2016, rows with population and GDP only).

## Step 1: Correlation
Measured how strongly medals, team size, population, GDP and GDP per person move together. Spearman correlation is used because medal counts are very skewed: a few countries win most medals, and Spearman is not distorted by that.

```python
d = panel_clean.copy()
d['log_medals'] = np.log1p(d['Total'])
d['log_athletes'] = np.log(d['athletes_sent'])

vars_ = ['Total', 'athletes_sent', 'population', 'gdp_usd', 'gdp_pc']
corr = d[vars_].corr(method='spearman').round(2)

print("Rows used:", len(d))
plt.figure(figsize=(6, 5))
sns.heatmap(corr, annot=True, cmap='Blues', vmin=0, vmax=1, fmt='.2f')
plt.title("Spearman correlation (1960-2016)")
plt.tight_layout()
plt.show()

corr['Total'].sort_values(ascending=False)
```

## Step 2: Regression
Fitted two models to explain medal counts. The medal count is transformed with log(1 + medals) and population and GDP are logged, because these variables span huge ranges.
- **Model 1:** population and GDP only
- **Model 2:** population, GDP and team size

R-squared shows the share of the variation in medals that the model explains.

```python
import statsmodels.formula.api as smf

m1 = smf.ols('log_medals ~ log_pop + log_gdp', data=d).fit()
m2 = smf.ols('log_medals ~ log_pop + log_gdp + log_athletes', data=d).fit()

for name, m in [("MODEL 1: population + GDP", m1),
                ("MODEL 2: population + GDP + team size", m2)]:
    print("=" * 55)
    print(name)
    print("=" * 55)
    res = pd.DataFrame({"coefficient": m.params.round(3),
                        "p_value": m.pvalues.round(4)})
    display(res)
    print("R-squared:", round(m.rsquared, 3), "| Observations:", int(m.nobs))
```

## Step 3: India's actual vs. expected medals
Used Model 1 (population and GDP) to predict how many medals a country of India's size and wealth would win at each Games, and compared that with what India actually won. The gap shows how much India under- or over-performs relative to its size and economy.

```python
d['predicted'] = np.expm1(m1.predict(d)).clip(lower=0)

ind_gap = (d[(d['NOC'] == 'IND') & (d['Year'] >= 1980)]
           [['Year', 'Total', 'predicted']]
           .rename(columns={'Total': 'actual'}).round(1))
ind_gap['gap'] = (ind_gap['actual'] - ind_gap['predicted']).round(1)
display(ind_gap)

print("Average gap 1980-2016:", ind_gap['gap'].mean().round(1))
```

```python
d16 = d[d['Year'] == 2016].copy()
x = np.log1p(d16['predicted'])
y = np.log1p(d16['Total'])

plt.figure(figsize=(7, 6))
plt.scatter(x, y, color='#8FA7C4', alpha=0.7, label='Other countries')
for noc, color in [('IND', '#C0504D'), ('CHN', '#1F3A5F'), ('USA', '#1F3A5F')]:
    r = d16[d16['NOC'] == noc]
    if len(r):
        plt.scatter(np.log1p(r['predicted']), np.log1p(r['Total']), color=color, s=90)
        plt.annotate(noc, (np.log1p(r['predicted'].iloc[0]), np.log1p(r['Total'].iloc[0])),
                     textcoords='offset points', xytext=(6, 4))
lim = max(x.max(), y.max())
plt.plot([0, lim], [0, lim], '--', color='grey', label='Actual = expected')
plt.title("Rio 2016: actual vs. expected medals\n(log scale, expected from population and GDP)")
plt.xlabel("Expected medals (log scale)")
plt.ylabel("Actual medals (log scale)")
plt.legend()
plt.tight_layout()
plt.show()
```

**Data note:** the model is fitted on 1960-2016 data. Applying it to 2020 and 2024 is an extrapolation, so treat those results as indicative.

## Step 1: Robustness check
Refitted the population-and-GDP model using only 2000-2016. If the coefficients and R-squared are similar to the full-period model, the Day 9 conclusions are not an artefact of old data, boycotts or changes in the Games.

```python
d_recent = d[d['Year'] >= 2000]
m1_recent = smf.ols('log_medals ~ log_pop + log_gdp', data=d_recent).fit()

comp = pd.DataFrame({
    'All Games (1960-2016)': [m1.params['log_pop'], m1.params['log_gdp'],
                              m1.rsquared, m1.nobs],
    '2000-2016 only': [m1_recent.params['log_pop'], m1_recent.params['log_gdp'],
                       m1_recent.rsquared, m1_recent.nobs],
}, index=['coefficient: log population', 'coefficient: log GDP',
          'R-squared', 'Observations']).round(3)
comp
```

## Step 2: Over- and under-performers
For every country, the model's residual is the difference between actual medals and the medals expected from population and GDP (on the log scale). Averaged over 2000-2016, a **positive** value means a country wins more than expected, and a **negative** value means it wins fewer.

Only countries with at least 10 million people and at least 4 Games with data are compared, so tiny countries do not dominate the list.

```python
d['resid'] = m1.resid
rec = d[d['Year'] >= 2000]

perf = (rec.groupby('NOC')
        .agg(games=('Year', 'nunique'),
             avg_resid=('resid', 'mean'),
             total_medals=('Total', 'sum'),
             avg_pop_millions=('population', lambda s: s.mean() / 1e6)))
big = perf[(perf['games'] >= 4) & (perf['avg_pop_millions'] >= 10)].copy()
big['rank'] = big['avg_resid'].rank(ascending=False, method='min').astype(int)

n = len(big)
print("Large countries compared:", n)
print("India's rank (1 = biggest over-performer):", big.loc['IND', 'rank'], "of", n)
z = (big.loc['IND', 'avg_resid'] - big['avg_resid'].mean()) / big['avg_resid'].std()
print("India's distance from the average large country: {:.1f} standard deviations".format(z))

print("\nBiggest over-performers:")
display(big.sort_values('avg_resid', ascending=False).head(8).round(2))
print("\nBiggest under-performers:")
display(big.sort_values('avg_resid').head(8).round(2))
```

```python
top15 = big.sort_values('avg_pop_millions', ascending=False).head(15).sort_values('avg_resid')
colors = ['#C0504D' if n_ == 'IND' else '#1F3A5F' for n_ in top15.index]

plt.figure(figsize=(8, 6))
plt.barh(top15.index, top15['avg_resid'], color=colors)
plt.axvline(0, color='grey', linewidth=1)
plt.title("The 15 most populous countries: medals vs. expectation, 2000-2016\n(left = fewer than expected, right = more)")
plt.xlabel("Average gap from expected medals (log scale)")
plt.tight_layout()
plt.show()
```
